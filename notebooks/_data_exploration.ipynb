```python
# Cell 6 — Specific business-rule checks
print("Negative fare_amount:", (df["fare_amount"] < 0).sum())
print("Negative trip_distance:", (df["trip_distance"] < 0).sum())
print("Negative total_amount:", (df["total_amount"] < 0).sum())
print("Dropoff before pickup:", (df["tpep_dropoff_datetime"] < df["tpep_pickup_datetime"]).sum())
print("Zero/negative passenger_count:", (df["passenger_count"] <= 0).sum())
```

```python
# Cell 7 — Categorical values
print(df["payment_type"].value_counts(dropna=False))
print(df["RatecodeID"].value_counts(dropna=False))
print(df["VendorID"].value_counts(dropna=False))
```

```python
# Cell 8 — Check if the correlated-null cluster corresponds to Flex Fare (payment_type == 0)

null_mask = df["passenger_count"].isnull()

print("Total rows with passenger_count null:", null_mask.sum())
print()
print("payment_type distribution WITHIN the null cluster:")
print(df.loc[null_mask, "payment_type"].value_counts(dropna=False))
print()
print("payment_type distribution OUTSIDE the null cluster:")
print(df.loc[~null_mask, "payment_type"].value_counts(dropna=False))
```

```python
# Cell 9 — Check VendorID within the null cluster
print("VendorID distribution WITHIN the null cluster:")
print(df.loc[null_mask, "VendorID"].value_counts(dropna=False))
print()
print("VendorID distribution OUTSIDE the null cluster:")
print(df.loc[~null_mask, "VendorID"].value_counts(dropna=False))
```
